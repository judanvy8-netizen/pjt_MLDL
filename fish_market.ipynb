### CSV에서 직접 만들어 볼 수 있습니다

```
Length2 → bream_length
Weight  → bream_weight
```

* list --> csv --> pandas --> 머신러닝

````python
import pandas as pd

fish = pd.read_csv('./data/Fish.csv')

bream = fish[fish['Species'] == 'Bream']
smelt = fish[fish['Species'] == 'Smelt']

bream_length = bream['Length2'].tolist()
bream_weight = bream['Weight'].tolist()

smelt_length = smelt['Length2'].tolist()
smelt_weight = smelt['Weight'].tolist()

print(len(bream_length))
print(len(smelt_length))
````

이제 다음 변수에 data를 추출해서 연결해 봅시다

```
bream_length
bream_weight
smelt_length
smelt_weight
```

````python
import matplotlib.pyplot as plt

plt.scatter(bream_length, bream_weight)
plt.scatter(smelt_length, smelt_weight)

plt.xlabel('length')
plt.ylabel('weight')

plt.show()
````
