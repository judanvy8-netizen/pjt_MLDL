```python
fish_length = [25.4, 26.3, 26.5, 29.0, 29.0, 29.7, 29.7, 30.0, 30.0, 30.7, 31.0, 31.0,
                31.5, 32.0, 32.0, 32.0, 33.0, 33.0, 33.5, 33.5, 34.0, 34.0, 34.5, 35.0,
                35.0, 35.0, 35.0, 36.0, 36.0, 37.0, 38.5, 38.5, 39.5, 41.0, 41.0, 9.8,
                10.5, 10.6, 11.0, 11.2, 11.3, 11.8, 11.8, 12.0, 12.2, 12.4, 13.0, 14.3, 15.0]
fish_weight = [242.0, 290.0, 340.0, 363.0, 430.0, 450.0, 500.0, 390.0, 450.0, 500.0, 475.0, 500.0,
                500.0, 340.0, 600.0, 600.0, 700.0, 700.0, 610.0, 650.0, 575.0, 685.0, 620.0, 680.0,
                700.0, 725.0, 720.0, 714.0, 850.0, 1000.0, 920.0, 955.0, 925.0, 975.0, 950.0, 6.7,
                7.5, 7.0, 9.7, 9.8, 8.7, 10.0, 9.9, 9.8, 12.2, 13.4, 12.2, 19.7, 19.9]
print(fish_length)
print(fish_weight)
print(type(fish_length))
print(type(fish_weight))
```

각 생선의 길이와 무게 --> zip(), for문 이용 --> 2차원 리스트 변환

```python
fish_data = [[l, w] for l, w in zip(fish_length, fish_weight)]
print(fish_data)
print(type(fish_data))

fish_target = [1]*35 + [0]*14
print(fish_target)
print(type(fish_target))
```

```python
from sklearn.neighbors import KNeighborsClassifier
kn = KNeighborsClassifier()  # kn 모델 생성
```

### 슬라이싱을 통해서 훈련 세트와 테스트 세트를 구성합니다

```python
# 훈련 세트로 입력값 중 0부터 34번 인덱스까지 사용
train_input = fish_data[:35]
# 훈련 세트로 타깃값 중 0부터 34번 인덱스까지 사용
train_target = fish_target[:35]
# 테스트 세트로 입력값 중 35번째부터 마지막 인덱스 까지 사용
test_input = fish_data[35:]
# 테스트 세트로 타깃값 중 35번째부터 마지막 인덱스 까지 사용
test_target = fish_target[35:]
```

훈련 세트로 --> 훈련 fit() 한다,   테스트 세트로 score(0 호출해 본다

```python
kn.fit(train_input, train_target)
kn.score(test_input, test_target)
```

### 왜 정확도 score() --> 0.0 인가요?

### 원인은 ?  샘플링 편향이 발행한 것입니다

“이렇게  마지막 14개를 테스트 세트로 떼어  놓으면 훈련 세트에는 빙어가 하나도 들어  있지 않잖아요. 빙어  없이 모델을 훈련하면 빙어를 올바르게 분류할 수가 없죠.”

### 그럼 어떻게 해야 하나요?

“훈련 세트와 테스트 세트를 나누려면 도미와 빙어가 골고루 섞이게 만들어야 해요.”

### 샘플링 편향 (sampling bias)

상식적으로 훈련하는 데이터와 테스트하는 데이터에는 도미와 빙어가 골고루 섞여 있어야 합니다. 일반적으로 훈련 세트와 테스트 세트에 샘플이 골고루 섞여 있지 않으면 샘플링이 한쪽으로 치우쳤 다는 의미로 샘플링 편향 이라고 부릅니다.
